## Problem Statement

The goal of this project is to segment customers into different groups
based on their purchasing behavior using clustering techniques.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
customers_df = pd.read_csv("/content/Mall_Customers.csv")
customers_df.head()
```

```python
customers_df.info()
```

```python
customers_df.describe()
```

```python
X = customers_df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']]
X.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

```python
kmeans = KMeans(n_clusters=5, random_state=42)
y_kmeans = kmeans.fit_predict(X_scaled)
```

```python
customers_df['Cluster'] = y_kmeans
customers_df.head()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x=customers_df['Annual Income (k$)'],
    y=customers_df['Spending Score (1-100)'],
    hue=customers_df['Cluster'],
    palette='viridis'
)
plt.title("Customer Segments")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.show()
```

```python
customers_df.groupby('Cluster').mean(numeric_only=True)
```

## Conclusion

Using K-Means clustering, customers were segmented into five distinct groups
based on age, annual income, and spending score.

This segmentation helps businesses understand different customer behaviors
and design targeted marketing strategies for each group.
